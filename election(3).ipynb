## STEP 2: CLEAN AND STANDARDIZE DATA

The different sources use inconsistent column names, data types, and formats.
Standardize them all into a common format.

```python

print("STEP 2: CLEAN AND STANDARDIZE DATA")


# BLOCK 2.1: Create voter ID extraction function
print("\n[2.1] Standardizing voter IDs...")

def extract_voter_id(val):
    """Extract numeric voter ID from various formats"""
    if pd.isna(val):
        return None
    val = str(val).strip()
    # Extract just the numeric part
    numeric = ''.join(c for c in val if c.isdigit())
    return int(numeric) if numeric else None

print("  Function created to extract numeric voter IDs")
print("  Example conversions:")
print(f"    '7004160' → {extract_voter_id('7004160')}")
print(f"    7004160 → {extract_voter_id(7004160)}")
print(f"    None → {extract_voter_id(None)}")
```

```python
# BLOCK 2.2: Clean Northgate data
print("\n[2.2] Cleaning Northgate data")
northgate = northgate_raw.copy()
northgate['voter_id'] = northgate['CertificateNumber'].apply(extract_voter_id)
northgate['vote_type'] = 'In-Person Early'
northgate['county'] = 'Northgate'
northgate['vote_date'] = pd.to_datetime(northgate['VoteDate'], errors='coerce')

print(f"  Records with valid IDs: {northgate['voter_id'].notna().sum()}/{len(northgate)}")
print(f"  Sample cleaned data:")
print(northgate[['voter_id', 'vote_type', 'county', 'vote_date']].head())
```

```python
# BLOCK 2.3: Clean Southfield data
print("\n[2.3] Cleaning Southfield data...")
southfield = southfield_raw.copy()
southfield['voter_id'] = southfield['voter_guid'].apply(extract_voter_id)
southfield['vote_type'] = 'In-Person Early'
southfield['county'] = 'Southfield'
southfield['vote_date'] = pd.to_datetime(southfield['date_voted'], errors='coerce')

print(f"  Records with valid IDs: {southfield['voter_id'].notna().sum()}/{len(southfield)}")
print(f"  Sample cleaned data:")
print(southfield[['voter_id', 'vote_type', 'county', 'vote_date']].head())
```

```python
# BLOCK 2.4: Clean mail ballot data
print("\n[2.4] Cleaning mail ballot data...")
mail = mail_raw.copy()
mail['voter_id'] = mail['VUID'].apply(extract_voter_id)
mail['vote_type'] = 'Mail-In'
mail['vote_date'] = pd.to_datetime(mail['Return Date'], errors='coerce')
# Note: county left empty, will be filled from registered voter roll

print(f"  Records with valid IDs: {mail['voter_id'].notna().sum()}/{len(mail)}")
print(f"  Sample cleaned data:")
print(mail[['voter_id', 'vote_type', 'vote_date']].head())
```

```python
# BLOCK 2.5: Combine all early voting records
print("\n[2.5] Combining early voting records...")

# First combine both in-person sources
early_inperson = pd.concat([
    northgate[['voter_id', 'vote_type', 'county', 'vote_date']],
    southfield[['voter_id', 'vote_type', 'county', 'vote_date']]
], ignore_index=True)

# Then add mail ballots
all_early = pd.concat([
    early_inperson,
    mail[['voter_id', 'vote_type', 'vote_date']].copy()
], ignore_index=True)

# Remove records with missing voter IDs
all_early = all_early.dropna(subset=['voter_id'])

print(f"  In-person early votes: {len(early_inperson)}")
print(f"  Mail votes: {len(mail)}")
print(f"  Total early votes: {len(all_early)}")
print(f"  Vote type breakdown:")
print(all_early['vote_type'].value_counts())
```

```python
# BLOCK 2.6: Match early votes with registered voter roll
print("\n[2.6] Matching early voters with registered voter roll...")

# Prepare registered voter data
voters = voters_raw[['SOS_VoterID', 'county', 'precinct_raw']].copy()
voters.columns = ['voter_id', 'county_reg', 'precinct']

# Merge early votes with voter information
runoff_votes = all_early.merge(voters, on='voter_id', how='left')

# Fill in county from registered roll where missing
runoff_votes['county'] = runoff_votes['county'].fillna(runoff_votes['county_reg'])
runoff_votes = runoff_votes.drop('county_reg', axis=1)

# Keep only records that matched (have precinct assignment)
runoff_votes_matched = runoff_votes.dropna(subset=['precinct'])

print(f"  Early votes matched to precincts: {len(runoff_votes_matched)}/{len(all_early)}")
print(f"  Unmatched votes: {len(all_early) - len(runoff_votes_matched)}")
print(f"  Data quality: {(len(runoff_votes_matched)/len(all_early)*100):.1f}% match rate")
print(f"\n  Votes by county:")
print(runoff_votes_matched['county'].value_counts())
```

```python
# BLOCK 2.7: Add May first-round voter tracking
print("\n[2.7] Adding first-round (May) voter tracking...")

# Load May first-round data
first_round = pd.read_csv('first_round_may.csv')
first_round['voter_id'] = first_round['voter_id'].apply(extract_voter_id)
first_round_voters = first_round[['voter_id']].drop_duplicates()
first_round_voters['voted_may'] = 1

print(f"  Total May voters: {len(first_round_voters)}")

# Merge with runoff data
runoff_votes_matched = runoff_votes_matched.merge(
    first_round_voters,
    on='voter_id',
    how='left'
)
runoff_votes_matched['voted_may'] = runoff_votes_matched['voted_may'].fillna(0)

may_in_runoff = runoff_votes_matched['voted_may'].sum()
print(f"  Runoff voters who also voted in May: {int(may_in_runoff)} ({may_in_runoff/len(runoff_votes_matched)*100:.1f}%)")
print(f"  New voters (May abstainers voting in runoff): {len(runoff_votes_matched) - int(may_in_runoff)} ({(1 - may_in_runoff/len(runoff_votes_matched))*100:.1f}%)")
```

## STEP 3: CALCULATE PRECINCT-LEVEL TURNOUT

Calculate turnout statistics for each precinct.

```python
print("\n" + "="*80)
print("STEP 3: CALCULATE PRECINCT-LEVEL TURNOUT")
print("="*80)

# BLOCK 3.1: Count registered voters by precinct
print("\n[3.1] Counting registered voters by precinct...")

registered_by_precinct = voters_raw.groupby('precinct_raw').size().reset_index(name='registered')
registered_by_precinct.columns = ['precinct', 'registered']

print(f"  Total precincts: {len(registered_by_precinct)}")
print(f"  Total registered voters: {registered_by_precinct['registered'].sum()}")
print(f"  Average precinct size: {registered_by_precinct['registered'].mean():.0f} voters")
print(f"  Precinct size range: {registered_by_precinct['registered'].min()} to {registered_by_precinct['registered'].max()}")
print(f"\n  Precincts by size:")
print(registered_by_precinct.sort_values('registered', ascending=False))
```

```python
# BLOCK 3.2: Count early voters by precinct
print("\n[3.2] Counting early voters by precinct...")

# Get unique voters per precinct (one vote per voter per precinct)
early_voters_unique = runoff_votes_matched.sort_values('vote_date').groupby(
    ['precinct', 'voter_id']
).first().reset_index()

early_by_precinct = early_voters_unique.groupby('precinct').size().reset_index(name='early_votes')

print(f"  Precincts with early votes: {len(early_by_precinct)}")
print(f"  Total early votes: {early_by_precinct['early_votes'].sum()}")
print(f"  Average per precinct: {early_by_precinct['early_votes'].mean():.1f}")
print(f"  Range: {early_by_precinct['early_votes'].min()} to {early_by_precinct['early_votes'].max()}")
```

```python
# BLOCK 3.3: Calculate turnout rates
print("\n[3.3] Calculating turnout rates...")

# Merge registered and early vote counts
turnout_data = registered_by_precinct.merge(
    early_by_precinct,
    on='precinct',
    how='left'
)

# Handle precincts with no early votes
turnout_data['early_votes'] = turnout_data['early_votes'].fillna(0).astype(int)

# Calculate turnout percentage
turnout_data['turnout_pct'] = (
    turnout_data['early_votes'] / turnout_data['registered'] * 100
).round(2)

# Sort by turnout for easy viewing
turnout_data = turnout_data.sort_values('turnout_pct', ascending=False)

# Calculate overall statistics
total_registered = turnout_data['registered'].sum()
total_early_votes = turnout_data['early_votes'].sum()
overall_turnout = (total_early_votes / total_registered * 100)

print(f"\n  OVERALL TURNOUT STATISTICS:")
print(f"  ─────────────────────────────")
print(f"  Total Registered: {total_registered:,}")
print(f"  Total Early Votes: {total_early_votes:,}")
print(f"  Overall Turnout Rate: {overall_turnout:.2f}%")
print(f"  Average Precinct Turnout: {turnout_data['turnout_pct'].mean():.2f}%")
print(f"  Median Precinct Turnout: {turnout_data['turnout_pct'].median():.2f}%")
print(f"  Std Dev: {turnout_data['turnout_pct'].std():.2f}%")
print(f"  Min - Max: {turnout_data['turnout_pct'].min():.2f}% - {turnout_data['turnout_pct'].max():.2f}%")
```

```python
# BLOCK 3.4: Add May participation analysis
print("\n[3.4] Analyzing May participation...")

may_participation = early_voters_unique.groupby('precinct')['voted_may'].sum().reset_index(name='may_voters')

turnout_data = turnout_data.merge(may_participation, on='precinct', how='left')
turnout_data['may_voters'] = turnout_data['may_voters'].fillna(0).astype(int)

# Calculate May participation rate among early voters
turnout_data['may_pct_of_early'] = (turnout_data['may_voters'] / turnout_data['early_votes'] * 100).round(1)
turnout_data.loc[turnout_data['early_votes'] == 0, 'may_pct_of_early'] = 0

print(f"\n  May Participation in Runoff:")
print(f"  ────────────────────────────")
print(f"  Total May voters in runoff: {turnout_data['may_voters'].sum()}")
print(f"  % of early runoff votes: {(turnout_data['may_voters'].sum() / total_early_votes * 100):.1f}%")
print(f"  Average % per precinct: {turnout_data['may_pct_of_early'].mean():.1f}%")
```

```python
# Display full precinct analysis
print("\n" + "="*80)
print("PRECINCT-LEVEL TURNOUT (Ranked by Turnout Rate)")
print("="*80)

display_df = turnout_data[['precinct', 'registered', 'early_votes', 'turnout_pct', 'may_voters']].copy()
display_df.columns = ['Precinct', 'Registered', 'Early Votes', 'Turnout %', 'May Voters']

print("\n" + display_df.to_string(index=False))

print("\n" + "="*80)
```

## STEP 4: PREPARE DATA FOR VISUALIZATION

Export results in formats suitable for the interactive HTML map and static reports.

```python
print("\n" + "="*80)
print("STEP 4: PREPARE DATA FOR VISUALIZATION")
print("="*80)

# BLOCK 4.1: Create GeoJSON precinct key mapping
print("\n[4.1] Creating precinct ID mappings...")

def precinct_to_pctkey(precinct_str):
    """Convert precinct_raw value to GeoJSON pctkey format"""
    try:
        p = str(precinct_str).strip()
        p_num = str(int(p)).zfill(3)
        return f"7000{p_num}"
    except:
        return None

turnout_data['pctkey'] = turnout_data['precinct'].apply(precinct_to_pctkey)

print(f"  Precinct ID mapping created")
print(f"  Example mappings:")
for idx in range(min(3, len(turnout_data))):
    row = turnout_data.iloc[idx]
    print(f"    {str(row['precinct']):3s} \u2192 {row['pctkey']}")
```

```python
# BLOCK 4.2: Export summary statistics to CSV
print("\n[4.2] Exporting summary statistics...")

summary_csv = turnout_data[[
    'precinct', 'pctkey', 'registered', 'early_votes', 'turnout_pct', 'may_voters'
]].copy()
summary_csv.columns = ['Precinct', 'Precinct_ID', 'Registered', 'Early_Votes', 'Turnout_%', 'May_Voters']
summary_csv = summary_csv.sort_values('Turnout_%', ascending=False)

summary_csv.to_csv('turnout_summary.csv', index=False)
print(f"  Saved: turnout_summary.csv")
print(f"\n  File contents:")
print(summary_csv.to_string(index=False))
```

```python
# BLOCK 4.3: Export data as JSON
print("\n[4.3] Exporting data as JSON...")

# Create data for JavaScript embedding in HTML
data_for_js = []
for idx, row in summary_csv.iterrows():
    data_for_js.append({
        'precinct': str(row['Precinct']),
        'pctkey': str(row['Precinct_ID']),
        'registered': int(row['Registered']),
        'early_votes': int(row['Early_Votes']),
        'turnout_pct': float(row['Turnout_%']),
        'may_voters': int(row['May_Voters'])
    })

with open('turnout_data.json', 'w') as f:
    json.dump(data_for_js, f, indent=2)
print(f"  Saved: turnout_data.json")
print(f"  Records: {len(data_for_js)}")
```
